Fill Nulls in (Calories_Burned) With Median

```python
df2 = df2[df2['Calories_Burned'] > 0]
```

```python
Q2 = df2.Calories_Burned.median()
Q2
```

```python
q1= np.quantile(df2.Calories_Burned, 0.25)
q1
```

```python
q3 = np.quantile(df2.Calories_Burned, 0.75 )
q3
```

```python
IQR = q3 - q1
IQR
```

```python
l = q1- (1.5 *IQR )
l
```

```python
u = q3 + (1.5 * IQR)
u
```

```python
out = df2[(df2.Calories_Burned < l) | 
         (df2.Calories_Burned > u)]

out
```

```python
df2 = df2[
    (df2['Calories_Burned'] <= u) &
    (df2['Calories_Burned'] >= l)
]
```

```python
df2['Calories_Burned'] = df2['Calories_Burned'].astype(int)
```
